## 5️⃣ 條件判斷 - 讓程式做決定 🧠

### if-else 基本語法

> 💬 **Prompt**：「Python的if-else怎麼寫？幫我寫一個最簡單的範例，加上註解說明」

> ⚠️ **重點**：Python用**縮排**（4個空格）來表示程式碼塊！

````python
age = 16
is_student = True

if age >= 18:
    print("你已經成年了！")
else:
    print("你還是學生呢！")

print(f"\n年齡：{age}，是學生嗎？{is_student}")
````

### 🌡️ 溫度判斷範例（if-elif-else）

> 💬 **Prompt**：「幫我寫一個Python程式，問我今天的溫度，然後根據溫度給建議：30度以上說好熱多喝水、25到30度說穿短袖就好、其他說涼爽適合出門」

````python
temperature = int(input("今天溫度是幾度？"))

if temperature >= 30:
    print("好熱啊！記得多喝水！🥵")
elif temperature >= 25:
    print("有點熱，穿短袖就好 👕")
else:
    print("涼爽舒適，適合出門走走！🚶‍♂️")
````

### ⚡ 邏輯運算符號（and / or / not）

> 💬 **Prompt**：「Python的and、or、not怎麼用？幫我寫三個範例，分別示範兩個條件都要滿足、任一條件滿足、還有反過來的邏輯」

````python
age = 17
is_student = True
score = 95
is_raining = False

# and (且) - 兩個條件都要滿足
if age >= 16 and is_student:
    print("你是高中生！")

# or (或) - 任一條件滿足即可
if score >= 90 or score <= 10:
    print("極端分數！")

# not (非) - 反過來
if not is_raining:
    print("今天適合運動！🏃")
````

## 🤖 設計你的第一個AI助手

### 什麼是Q&A聊天機器人？
一個能回答問題的程式！就像你問它：
- 「今天天氣如何？」→ 它回答你
- 「Python是什麼？」→ 它解釋給你聽

### 我們的策略：**關鍵字匹配** 🔑
- 早期的聊天室AI與客服AI所使用的模式！

> 💬 **Prompt**：「幫我寫一個簡單的Q&A程式，問我問題，如果問題裡有『天氣』就回答今天天氣不錯、如果有python就解釋Python是什麼、其他都說不確定」

````python
question = input("你想問什麼？")

if "天氣" in question:
    print("今天天氣不錯！☀️")
elif "python" in question.lower():
    print("Python是一種程式語言！🐍")
else:
    print("我不太確定呢，再想想看～ 🤔")
````

### 完整的Q&A聊天機器人程式碼

> 💬 **Prompt**：「幫我寫一個Q&A聊天機器人，要有無限迴圈讓對話持續進行、輸入『離開』就結束、能回答打招呼、天氣、Python介紹、問名字這些問題、不認識的問題要說還在學習中」

````python
print("=" * 40)
print("   🤖 AI小助手 - Q&A聊天機器人")
print("=" * 40)
print("輸入 '離開' 結束對話\n")

while True:
    question = input("你：").strip()
    
    if question == "離開":
        print("AI：拜拜！下次再聊～ 👋")
        break
    
    elif "你好" in question or "哈囉" in question:
        print("AI：嗨！有什麼可以幫你的嗎？😊")
    
    elif "天氣" in question:
        print("AI：今天天氣不錯，適合出門走走！☀️")
    
    elif "python" in question.lower():
        print("AI：Python是一種易學的程式語言，廣泛用於AI和資料分析！🐍")
    
    elif "名字" in question or "叫什麼" in question:
        print("AI：我是AI小助手，很高興認識你！🤖")
    
    else:
        print("AI：嗯...這個問題我還在學習中呢 🤔 試試問我天氣或Python吧！")
````

### 💡 程式碼解析

| 語法 | 說明 |
|------|------|
| `while True:` | 無限迴圈，讓對話持續進行 |
| `.strip()` | 去除輸入前後的多餘空白 |
| `"關鍵字" in question` | 檢查問題中是否包含關鍵字 |
| `.lower()` | 轉成小寫，讓匹配不分大小寫 |
| `break` | 跳出迴圈，結束程式 |

> 🎯 **核心概念**：用 if-elif-else 判斷使用者輸入的關鍵字，給出對應回答！
